# Multisource AI Content Intelligence Platform

## Baseline Implementation

This notebook implements the baseline content summarization pipeline
for the Multisource AI Content Intelligence Platform.

### Baseline capabilities

- Accept a website URL
- Accept a YouTube URL
- Extract textual content
- Send the extracted content to an LLM
- Generate a concise summary

### Purpose

This baseline will later be extended into a complete AI Content
Intelligence Platform with document processing, embeddings,
vector search, RAG-based question answering, and content analytics.

```python
import os
import validators

from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_groq import ChatGroq
```

```python
load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")

if not groq_api_key:
    raise ValueError("GROQ_API_KEY was not found. Check your .env file.")

```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    groq_api_key=groq_api_key,
    temperature=0
)
```

```python
response = llm.invoke(
    "Explain what an LLM is in 3 simple sentences."
)

print(response.content)
```

## Baseline Summarization

The baseline pipeline uses an LLM to summarize extracted content into a
concise and structured summary.

The summarization process focuses on:

- The main idea
- Important facts
- Key points
- The overall conclusion

This provides the initial summarization capability that will later be
extended with multiple content sources, document processing, retrieval,
and content intelligence features.

```python
prompt_template = """
You are a helpful content summarization assistant.

Summarize the following content in approximately 300 words.

Focus on:
- The main idea
- Important facts
- Key points
- The overall conclusion

Content:
{text}
"""

prompt = PromptTemplate(
    template=prompt_template,
    input_variables=["text"]
)
```

```python
sample_text = """
Artificial intelligence is a field of computer science concerned with
creating systems that can perform tasks that normally require human
intelligence. Machine learning is a major part of artificial
intelligence where systems learn patterns from data. Deep learning
uses neural networks with multiple layers to learn complex patterns.
Generative AI extends these capabilities by creating new content such
as text, images, audio and code. Large language models are a type of
generative AI model trained on large amounts of text and can generate,
summarize and transform natural language.
"""
```

```python
formatted_prompt = prompt.format(text=sample_text)

response = llm.invoke(formatted_prompt)

print(response.content)
```
